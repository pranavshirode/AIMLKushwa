```python

import math
import heapq

# Coordinates of 8 locations
positions = {
    "A": (0, 0),
    "B": (2, 0),
    "C": (1, 2),
    "D": (4, 0),
    "E": (3, 2),
    "F": (6, 0),
    "G": (5, 2),
    "H": (7, 2)
}

# Create weighted graph
graph = {city: {} for city in positions}

edges = [
    ("A", "B"), ("A", "C"), ("B", "C"),
    ("B", "D"), ("C", "E"), ("D", "E"),
    ("D", "F"), ("E", "G"), ("F", "G"),
    ("F", "H"), ("G", "H")
]

for u, v in edges:
    distance = math.dist(positions[u], positions[v])
    graph[u][v] = distance
    graph[v][u] = distance

# Heuristic function
def heuristic(city, goal):
    return math.dist(positions[city], positions[goal])

# Greedy Best-First Search
def greedy_search(start, goal):
    queue = [(heuristic(start, goal), start)]
    parent = {start: None}
    cost = {start: 0}
    explored = []

    while queue:
        h, city = heapq.heappop(queue)

        if city in explored:
            continue

        explored.append(city)

        if city == goal:
            path = []
            while city is not None:
                path.append(city)
                city = parent[city]

            path.reverse()
            return path, cost[goal], explored

        for neighbor, distance in graph[city].items():
            if neighbor not in parent:
                parent[neighbor] = city
                cost[neighbor] = cost[city] + distance
                heapq.heappush(
                    queue, (heuristic(neighbor, goal), neighbor)
                )

    return None, 0, explored


start = input("Enter source location (A-H): ").upper()
goal = input("Enter destination location (A-H): ").upper()

if start not in graph or goal not in graph:
    print("Invalid location!")
else:
    path, total_cost, explored = greedy_search(start, goal)

    if path:
        print("Selected route:", " -> ".join(path))
        print("Total path cost:", round(total_cost, 2))
        print("Nodes explored:", len(explored))
    else:
        print("No route found.")
```

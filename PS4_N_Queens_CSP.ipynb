```python

# N-Queens using CSP and Backtracking

N = int(input("Enter N (minimum 4): "))

board = [-1] * N

# Check whether a queen can be placed
def is_safe(row, col):
    for i in range(row):
        if board[i] == col:
            return False

        if abs(board[i] - col) == abs(i - row):
            return False

    return True

# Backtracking function
def solve(row):
    if row == N:
        return True

    for col in range(N):
        if is_safe(row, col):
            board[row] = col

            if solve(row + 1):
                return True

            board[row] = -1

    return False

# Solve the problem
if N < 4:
    print("N must be at least 4")
else:
    if solve(0):
        print("Solution found:")

        for row in range(N):
            for col in range(N):
                if board[row] == col:
                    print("Q", end=" ")
                else:
                    print(".", end=" ")
            print()

        # Verify the solution
        valid = True

        for i in range(N):
            for j in range(i + 1, N):
                if board[i] == board[j]:
                    valid = False

                if abs(board[i] - board[j]) == abs(i - j):
                    valid = False

        if valid:
            print("Verification: No queens are attacking each other.")
        else:
            print("Verification failed.")

    else:
        print("No solution found.")
```

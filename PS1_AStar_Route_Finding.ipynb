```python

import math
import heapq

# Coordinates of 8 cities on a sample map
positions = {
    "A": (0, 0),
    "B": (2, 0),
    "C": (1, 2),
    "D": (4, 0),
    "E": (3, 2),
    "F": (6, 0),
    "G": (5, 2),
    "H": (7, 2)
}

# Create a weighted graph
graph = {city: {} for city in positions}

edges = [
    ("A", "B"), ("A", "C"), ("B", "C"),
    ("B", "D"), ("C", "E"), ("D", "E"),
    ("D", "F"), ("E", "G"), ("F", "G"),
    ("F", "H"), ("G", "H")
]

for u, v in edges:
    distance = math.dist(positions[u], positions[v])
    graph[u][v] = distance
    graph[v][u] = distance

# Heuristic: straight-line distance to the destination
def heuristic(city, goal):
    return math.dist(positions[city], positions[goal])

# A* Search
def a_star(start, goal):
    queue = [(heuristic(start, goal), start)]
    cost = {start: 0}
    parent = {start: None}
    explored = []

    while queue:
        f, city = heapq.heappop(queue)

        if city in explored:
            continue

        explored.append(city)

        if city == goal:
            path = []
            while city is not None:
                path.append(city)
                city = parent[city]

            path.reverse()
            return path, cost[goal], explored

        for neighbor, distance in graph[city].items():
            new_cost = cost[city] + distance

            if neighbor not in cost or new_cost < cost[neighbor]:
                cost[neighbor] = new_cost
                parent[neighbor] = city
                f = new_cost + heuristic(neighbor, goal)
                heapq.heappush(queue, (f, neighbor))

    return None, 0, explored


start = input("Enter source city (A-H): ").upper()
goal = input("Enter destination city (A-H): ").upper()

if start not in graph or goal not in graph:
    print("Invalid city!")
else:
    path, total_cost, explored = a_star(start, goal)

    if path:
        print("Selected route:", " -> ".join(path))
        print("Total path cost:", round(total_cost, 2))
        print("Nodes explored:", len(explored))
    else:
        print("No route found.")
```

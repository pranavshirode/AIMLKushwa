```python

# Sudoku Solver using CSP and Backtracking

# Enter 9 rows, using 0 for empty cells
board = []

print("Enter Sudoku row by row (use 0 for empty cells):")
for i in range(9):
    row = list(map(int, input().strip()))
    board.append(row)

# Check row, column and 3x3 subgrid constraints
def is_safe(board, row, col, num):
    for i in range(9):
        if board[row][i] == num:
            return False

        if board[i][col] == num:
            return False

    start_row = row - row % 3
    start_col = col - col % 3

    for i in range(3):
        for j in range(3):
            if board[start_row + i][start_col + j] == num:
                return False

    return True

# Check whether the initial puzzle is valid
def valid_grid(board):
    for row in range(9):
        for col in range(9):
            num = board[row][col]

            if num != 0:
                board[row][col] = 0
                safe = is_safe(board, row, col, num)
                board[row][col] = num

                if not safe:
                    return False

    return True

# Solve using backtracking
def solve(board):
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:
                for num in range(1, 10):
                    if is_safe(board, row, col, num):
                        board[row][col] = num

                        if solve(board):
                            return True

                        board[row][col] = 0

                return False

    return True

# Validate and solve the puzzle
if valid_grid(board) and solve(board):
    print("Solved Sudoku:")

    for row in board:
        print(*row)

    print("Solution is valid.")
else:
    print("Sudoku has no valid solution.")
```

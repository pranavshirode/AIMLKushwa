```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Load Mall Customers dataset
df = pd.read_csv("Mall_Customers.csv")
print(df.head())

# 2. Select Annual Income and Spending Score
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

# 3. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Apply Elbow Method
wcss = []

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    wcss.append(model.inertia_)

# Plot Elbow Graph
plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()

# 5. Train K-Means with K = 5
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to dataset
df["Cluster"] = clusters

# 6. Calculate number of customers in each cluster
print("\nNumber of Customers in Each Cluster:")
print(df["Cluster"].value_counts().sort_index())

# Get centroids in original scale
centroids = scaler.inverse_transform(kmeans.cluster_centers_)

# 7. Visualize customer clusters and centroids
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x=X["Annual Income (k$)"],
    y=X["Spending Score (1-100)"],
    hue=df["Cluster"],
    palette="Set1",
    s=60
)

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker="X",
    s=250,
    color="black",
    label="Centroids"
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")
plt.title("Mall Customer Clustering using K-Means")
plt.legend()
plt.show()

# Display customer segment characteristics
print("\nCustomer Segment Characteristics:")
print(df.groupby("Cluster")[
    ["Annual Income (k$)", "Spending Score (1-100)"]
].mean().round(2))
```

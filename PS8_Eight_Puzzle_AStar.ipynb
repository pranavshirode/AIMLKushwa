```python

import heapq

start = (1, 2, 3, 4, 0, 6, 7, 5, 8)
goal  = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def h(state):
    distance = 0
    for i in range(9):
        if state[i] != 0:
            j = goal.index(state[i])
            distance += abs(i//3 - j//3) + abs(i%3 - j%3)
    return distance

def moves(state):
    result = []
    z = state.index(0)
    for n in [z-3, z+3, z-1, z+1]:
        if 0 <= n < 9 and (abs(n-z) == 3 or n//3 == z//3):
            s = list(state)
            s[z], s[n] = s[n], s[z]
            result.append(tuple(s))
    return result

pq = [(h(start), 0, start, [start])]
visited = set()
explored = 0

while pq:
    f, g, state, path = heapq.heappop(pq)

    if state in visited:
        continue
    visited.add(state)
    explored += 1

    if state == goal:
        for s in path:
            print(s[:3], "\n", s[3:6], "\n", s[6:], "\n")
        print("Moves:", len(path)-1)
        print("States explored:", explored)
        break

    for nxt in moves(state):
        if nxt not in visited:
            heapq.heappush(pq, (g+1+h(nxt), g+1, nxt, path+[nxt]))
else:
    print("No solution found")
```

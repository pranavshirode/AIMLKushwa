```python

exams = ["Maths", "Physics", "Chemistry", "CN", "DBMS", "AI"]
slots = [1, 2, 3, 4]

students = {
    "S1": ["Maths", "Physics"],
    "S2": ["Maths", "Chemistry"],
    "S3": ["Physics", "CN"],
    "S4": ["Chemistry", "DBMS"],
    "S5": ["CN", "DBMS", "AI"],
    "S6": ["Maths", "AI"]
}

# Find conflicting exams
conflicts = {e: set() for e in exams}
for subjects in students.values():
    for a in subjects:
        for b in subjects:
            if a != b:
                conflicts[a].add(b)

timetable = {}

def solve(i):
    if i == len(exams):
        return True

    e = exams[i]
    for s in slots:
        if all(timetable.get(c) != s for c in conflicts[e]):
            timetable[e] = s
            if solve(i + 1):
                return True
            del timetable[e]

    return False

if solve(0):
    for e in exams:
        print(e, "-> Slot", timetable[e])
    print("Verification:", all(
        timetable[a] != timetable[b]
        for a in exams for b in conflicts[a]
    ))
else:
    print("No valid timetable")
```
